# Часть 1. Переменные, числа и строки

**Переменная** — имя, привязанное к значению. Тип значения Python определяет сам, но знать его
обязан программист: от типа зависит, что с этим значением можно делать.

Четыре типа, которые нужны сегодня:

| Тип | Что это | Пример |
|---|---|---|
| `int` | целое число | `qty = 2` |
| `float` | число с дробной частью | `price = 320.50` |
| `str` | строка (текст) | `item = "Латте"` |
| `bool` | истина или ложь | `is_big = True` |

Главное правило дня: **всё, что прочитано из файла, — это строка `str`**, даже если выглядит
как число. `"320"` и `320` — разные вещи, и складывать их Python откажется.

````python
qty = 2                 # int
price = 320.50          # float
item = "Латте"          # str
is_big = False          # bool

print(qty, type(qty))
print(price, type(price))
print(item, type(item))
print(is_big, type(is_big))

# Арифметика
print("сумма позиции:", qty * price)
print("деление:      ", 7 / 2)     # всегда float
print("целая часть:  ", 7 // 2)    # отбрасывает дробь
print("остаток:      ", 7 % 2)     # пригодится для проверки чётности
print("степень:      ", 2 ** 10)

# Преобразование типов
print(int("320"), type(int("320")))       # строка -> число
print(str(320) + " руб.")                 # число -> строка
print(float("320.50") * 2)
````

## Строки: то, чем мы будем чистить данные

Из файла всё приходит текстом, поэтому строковые методы — главный инструмент аналитика
на этом этапе. Метод вызывается через точку и **возвращает новую строку**, а не меняет старую.

| Метод | Что делает |
|---|---|
| `.strip()` | убирает пробелы по краям |
| `.lower()` / `.upper()` | нижний / верхний регистр |
| `.replace(a, b)` | заменяет одно на другое |
| `.split(";")` | режет строку по разделителю, возвращает список |
| `.capitalize()` | первая буква заглавная, остальные строчные |
| `len(s)` | длина строки |

````python
s = "  ЛаТтЕ  "

print(repr(s))                      # repr показывает пробелы и кавычки
print(repr(s.strip()))
print(repr(s.strip().lower()))
print(repr(s.strip().lower().capitalize()))

# Разбор строки из файла
line = "0007;2025-03-04;Капучино;2;300.00"
parts = line.split(";")
print(parts)
print("товар:", parts[2], "| количество:", parts[3])
print("частей в строке:", len(parts))

# Срезы: берём кусок строки по позициям
date = parts[1]
print("год:", date[0:4], "| месяц:", date[5:7])
````

# Часть 2. Логические выражения и условия

**Логическое выражение** — то, что вычисляется в `True` или `False`.

| Оператор | Смысл |
|---|---|
| `==`, `!=` | равно, не равно |
| `>`, `<`, `>=`, `<=` | сравнение |
| `and`, `or`, `not` | и, или, не |
| `in` | входит ли элемент в список / подстрока в строку |

**Две ловушки, на которых спотыкаются все:**

- `"10" == 10` даёт `False` — строка никогда не равна числу.
- `"10" > 5` не даёт ничего: программа падает с `TypeError`. Сравнивать можно только сравнимое.

Отсюда правило: сначала приводим тип, потом сравниваем.

````python
qty = 3
price = 320.0
total = qty * price

print(total > 500)
print(qty > 0 and price > 0)          # обе части истинны
print(qty > 5 or price > 100)         # хотя бы одна истинна
print(not qty > 5)

print("Латте" in ["Латте", "Раф"])    # элемент в списке
print("n/a" in "n/a")                 # подстрока в строке
print("10" == 10)                     # ловушка: разные типы

# Условие: ветка выбирается по первому истинному условию
if total >= 1000:
    category = "крупный"
elif total >= 500:
    category = "средний"
else:
    category = "малый"

print(f"Заказ на {total:.2f} руб. — {category}")
````

# Часть 3. Циклы и последовательности

**Список** (`list`) — упорядоченная последовательность значений. Нумерация с нуля;
отрицательные индексы считают с конца.

**Цикл `for`** проходит по последовательности и выполняет тело для каждого элемента.
Приём, который лежит в основе любой агрегации, — **накопитель**: переменная объявляется
до цикла и обновляется на каждом шаге.

````python
prices = [320.0, 180.0, 420.0, 300.0, 240.0]

print("первый:", prices[0], "| последний:", prices[-1])
print("первые три:", prices[:3])
print("длина:", len(prices))

# Накопитель: складываем всё вручную
total = 0                      # накопитель ДО цикла
for p in prices:
    total = total + p          # можно короче: total += p
print("сумма:", total)

# enumerate даёт номер элемента вместе со значением
for i, p in enumerate(prices, start=1):
    print(f"  позиция {i}: {p:.2f}")

# range — последовательность чисел
for i in range(3):
    print("шаг", i)

# continue пропускает шаг, break прерывает цикл
for p in prices:
    if p < 250:
        continue               # дешёвые пропускаем
    print("дорогая позиция:", p)
````

## Словарь: главный инструмент группировки

**Словарь** (`dict`) хранит пары «ключ — значение». Для аналитика это способ накапливать
итоги по категориям: ключ — название товара, значение — выручка по нему.

Ключевой приём — `d.get(key, 0)`: вернуть текущее значение, а если ключа ещё нет, вернуть 0.
Именно так и строится накопление по группам, то есть то, что pandas делает через `groupby`.

````python
sales = [
    ("Латте", 640.0),
    ("Раф", 380.0),
    ("Латте", 320.0),
    ("Чай", 200.0),
    ("Раф", 760.0),
]

revenue_by_item = {}                     # пустой словарь-накопитель

for item, total in sales:
    revenue_by_item[item] = revenue_by_item.get(item, 0) + total

print(revenue_by_item)
print()

for item, total in revenue_by_item.items():
    print(f"{item:<12} {total:>8.2f} руб.")

# Сортировка словаря по значению, по убыванию
top = sorted(revenue_by_item.items(), key=lambda pair: pair[1], reverse=True)
print()
print("Лидер:", top[0])
````

# Часть 4. Обработка исключений

`int("две")` не может дать число — Python **выбрасывает исключение** `ValueError`, и программа
останавливается. Для аналитического скрипта это плохой сценарий: одна испорченная строка из
двадцати тысяч убивает весь расчёт, причём после сорока минут работы.

Правильное поведение: проблемную строку пропустить, **посчитать её в счётчике** и продолжить.

```
try:
    ...код, который может упасть...
except ValueError:
    ...что делать, если упал...
```

Ловите **конкретный** тип ошибки. Голый `except:` перехватывает вообще всё, включая ваши
собственные опечатки, и превращает отладку в мучение.

````python
values = ["2", "3", "две", "", "4", "n/a", " 5 "]

good = 0
bad = 0

for v in values:
    try:
        number = int(v)
    except ValueError:
        print(f"  не смог разобрать: {v!r}")
        bad += 1
        continue          # переходим к следующему значению
    good += 1
    print(f"  разобрал: {v!r} -> {number}")

print()
print(f"успешно: {good}, ошибок: {bad}")

# int() умеет игнорировать пробелы по краям, но не запятые:
print(int(" 5 "))
try:
    print(float("320,00"))
except ValueError as e:
    print("а вот запятая ломает float:", e)
````

Напишите функцию `parse_price(text)`, которая превращает текст в число:

- умеет обрабатывать `"320.00"`, `"320,00"`, `" 320 "`, `"1 250,50"`;
- если превратить в число невозможно (`"n/a"`, `""`), возвращает `None` вместо падения.

Функция объявляется через `def`, результат отдаётся через `return`.

# Часть 5. Мини-проект: отчёт по продажам

Запустите ячейку ниже — она создаст файл `data/sales_raw.csv`. Это выгрузка продаж кофейни
за март 2025: 240 строк, разделитель `;`, первая строка — заголовок.

**Формат строки:** `order_id;date;item;qty;price`

**Что с данными не так** (специально, как в жизни):

| Дефект | Пример | Что делать |
|---|---|---|
| лишние пробелы и регистр в названии | `  латте ` | привести к единому виду |
| запятая вместо точки в цене | `320,00` | заменить и разобрать |
| количество текстом или пустое | `две`, `` | пропустить строку, посчитать в счётчике |
| цена не выгрузилась | `n/a` | пропустить строку, посчитать в счётчике |
| потеряна колонка | в строке 4 части вместо 5 | пропустить строку, посчитать в счётчике |
| отрицательное количество | `-1` | пропустить: это возврат, а не продажа |

````python
# ЯЧЕЙКА-ГЕНЕРАТОР ДАННЫХ. Запустите её (Shift+Enter), менять ничего не нужно.
# Создаёт data/sales_raw.csv — выгрузку продаж кофейни за март 2025.
# Данные намеренно "грязные": примерно такие выгрузки и приходят аналитику в реальности.

import random
from pathlib import Path

random.seed(7)

menu = [
    ("Латте", 320), ("Капучино", 300), ("Американо", 220), ("Раф", 380),
    ("Эспрессо", 180), ("Круассан", 240), ("Чизкейк", 420), ("Сэндвич", 390),
    ("Какао", 280), ("Чай", 200),
]

lines = ["order_id;date;item;qty;price"]

for i in range(1, 241):
    order_id = f"{i:04d}"
    date = f"2025-03-{random.randint(1, 31):02d}"
    name, price = random.choice(menu)
    qty = random.choices([1, 2, 3, 4], weights=[55, 28, 12, 5])[0]

    # название товара: где-то лишние пробелы, где-то другой регистр
    r = random.random()
    if r < 0.12:
        item_str = "  " + name.lower() + " "
    elif r < 0.18:
        item_str = name.upper()
    else:
        item_str = name

    qty_str = str(qty)
    price_str = f"{price}.00"

    # дефекты выгрузки
    d = random.random()
    if d < 0.04:
        qty_str = ""                                   # количество не выгрузилось
    elif d < 0.07:
        qty_str = random.choice(["две", "n/a", "-"])    # количество текстом
    elif d < 0.13:
        price_str = f"{price},00"                       # запятая вместо точки
    elif d < 0.16:
        price_str = "n/a"                               # цена не выгрузилась
    elif d < 0.19:
        qty_str = "-1"                                  # возврат записан как минус
    elif d < 0.22:
        lines.append(f"{order_id};{date};{item_str};{qty_str}")   # потеряна колонка
        continue
    elif d < 0.25:
        qty_str = f" {qty} "                            # лишние пробелы

    lines.append(f"{order_id};{date};{item_str};{qty_str};{price_str}")

Path("data").mkdir(exist_ok=True)
Path("data/sales_raw.csv").write_text("\n".join(lines), encoding="utf-8")

print("Готово. Создан файл data/sales_raw.csv")
print("Строк вместе с заголовком:", len(lines))
print()
print("Первые 8 строк файла:")
for line in lines[:8]:
    print("   ", line)
````

### Задание 8. Главный цикл

Соберите всё, что было на паре, в один цикл. Порядок проверок важен: сначала структура строки,
потом типы, потом смысл значений.

Для каждой строки:

1. разрезать по `;`;
2. если частей не 5 — увеличить `bad_structure` и перейти к следующей строке;
3. попытаться получить число из количества и цены; при `ValueError` — увеличить `bad_number` и перейти дальше;
4. если количество или цена не больше нуля — увеличить `bad_logic` и перейти дальше;
5. иначе: посчитать стоимость позиции, добавить её к общей выручке, увеличить `good`
   и накопить выручку по товару в словаре (название — привести к единому виду).

````python
good = 0
bad_structure = 0
bad_number = 0
bad_logic = 0

revenue = 0.0
revenue_by_item = {}

for line in rows:
    parts = line.split(";")

    # 2) проверка структуры
    
    if len(parts) != 5:
        bad_structure += 1
        continue

    # 3) разбор чисел (try / except ValueError)
    try:
        num1 = int(parts[3])
        # num2 = float(parts[4])
    except ValueError:
        bad_number += 1
        continue

    try:
        num2 = float(parts[4].replace(',', '.'))
    except ValueError:
        bad_number += 1
        continue

    # 4) проверка смысла значений
    if num1 <= 0:
        bad_logic += 1
    # if num1 < 0 or num2 < 0:
    #     total = num2 * num1
    #     item = parts[2].strip().lower().capitalize()
    #     revenue_by_item[item] = revenue_by_item.get(item, 0) + total

    # 5) накопление
    else:
        total = num1 * num2
        good += 1
        item = parts[2].strip().lower().capitalize()
        revenue_by_item[item] = revenue_by_item.get(item, 0) + total

for items, prices in revenue_by_item.items():
    revenue += prices 
        
print("успешно разобрано:", good)
print("битая структура:  ", bad_structure)
print("не число:         ", bad_number)
print("нелогичные данные:", bad_logic)
print(f"выручка: {revenue:.2f}")
````

## Бонус: то же самое на pandas

Теперь, когда понятно, что происходит внутри, посмотрите на то же самое библиотекой.
Обратите внимание: pandas не делает данные чистыми сам — решение о том, что считать
браком, всё равно принимаете вы. Он лишь избавляет от ручного цикла.

````python
import pandas as pd

df = pd.read_csv("data/sales_raw.csv", sep=";", on_bad_lines="skip")

df["qty"] = pd.to_numeric(df["qty"], errors="coerce")           # непреобразуемое -> NaN
df["price"] = pd.to_numeric(
    df["price"].astype(str).str.replace(",", ".", regex=False), errors="coerce"
)
df["item"] = df["item"].str.strip().str.lower()

clean = df[(df["qty"] > 0) & (df["price"] > 0)].copy()
clean["total"] = clean["qty"] * clean["price"]

print("строк после очистки:", len(clean))
print(f"выручка: {clean['total'].sum():.2f}")
print()
print(clean.groupby("item")["total"].sum().sort_values(ascending=False).head(3))
````
